```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BMIState(TypedDict):
    weight_kg: float
    height_m: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    weight = state['weight_kg']
    height = state['height_m']

    bmi = weight/(height**2)

    state['bmi'] = round(bmi, 2)      # partial update 

    return state



def label_bmi(state: BMIState) -> BMIState:
    if state['bmi'] < 18.5:
        state['category'] = 'Underweight'
    elif state['bmi'] <= 24.9:
        state['category'] = 'Healthy weight'
    elif state['bmi'] <= 29.9:
        state['category'] = 'Overweight'
    elif state['bmi'] >= 30:
        state['category'] = 'Obese'

    return state
```

```python
# define graph
graph = StateGraph(BMIState)


# add nodes
# .add_node(node name, python function)
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('label_bmi', label_bmi)


# add edges
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'label_bmi')
graph.add_edge('label_bmi', END)


# compile graph
workflow = graph.compile()      # always take initial state and return final state
```

```python
# execute graph
initial_state = {'weight_kg': 60, 'height_m': 1.79832}
final_state = workflow.invoke( initial_state )

print(final_state)
```
