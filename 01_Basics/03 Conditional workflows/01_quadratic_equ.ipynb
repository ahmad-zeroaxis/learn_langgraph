```python
from langgraph.graph import START, END, StateGraph
from typing import TypedDict, Literal
```

```python
class QuadraticState(TypedDict):
    a: int
    b: int
    c: int

    equation: str
    discriminant: float
    result: str
```

```python
def show_equation(state: QuadraticState):

    # abs() means absolute value. 
    # It removes the negative sign from a number.  e.g., abs(-5) = 5      abs(5) = 5

    a = state['a']
    b = state['b']
    c = state['c']

    equation = f"{a}x²"

    if b >= 0:
        equation += f" + {b}x"
    else:
        equation += f" - {abs(b)}x"

    if c >= 0:
        equation += f" + {c}"
    else:
        equation += f" - {abs(c)}"

    return {'equation': equation}



def calculate_equation(state: QuadraticState):
    # d = b² - 4ac 
    discriminant = (state['b']**2) - (4*state['a']*state['c'])

    return {'discriminant': discriminant}





def two_real_roots(state: QuadraticState):      # d > 0
    # root1 = -b + √d / 2a 
    root1 = ( (-state['b']) + (state['discriminant']**0.5) ) / (2*state['a'])

    # root1 = -b - √d / 2a 
    root2 = ( (-state['b']) - (state['discriminant']**0.5) ) / (2*state['a'])


    return {'result': f"Two real roots: {root1} and {root2}"}



def one_real_repeated_root(state: QuadraticState):      # d = 0
    # root1 = -b / 2a 
    root = (-state['b']) / (2*state['a'])


    return {'result': f"One real repeated root: {root} occutes twice "}



def no_real_roots(state: QuadraticState):       # d < 0
    result = "No real toots"

    return {'result': result}




def check_condition(state: QuadraticState) -> Literal["two_real_roots", "one_real_repeated_root", "no_real_roots"]:
    d = state['discriminant']

    if d > 0:
        return "two_real_roots"
    elif d == 0:
        return "one_real_repeated_root"
    else:   # d < 0
        return "no_real_roots"
```

```python
graph = StateGraph(QuadraticState)


graph.add_node('show_equation', show_equation)
graph.add_node('calculate_equation', calculate_equation)
graph.add_node('two_real_roots', two_real_roots)
graph.add_node('one_real_repeated_root', one_real_repeated_root)
graph.add_node('no_real_roots', no_real_roots)



graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation', 'calculate_equation')
graph.add_conditional_edges('calculate_equation', check_condition)
graph.add_edge('two_real_roots', END)
graph.add_edge('one_real_repeated_root', END)
graph.add_edge('no_real_roots', END)


workflow = graph.compile()
```

```python
initial_state = {
    'a': 2,
    'b': 3,
    'c': -5
}

final_state = workflow.invoke(initial_state)
```
